```python
#Version 1

import numpy as np
def calcul_moyenne():
    Aa=int(input("Donner le coefficient de ... dans UE1"))
    Ba=int(input("Donner le coefficient de ... dans UE1"))
    Ca=int(input("Donner le coefficient de ... dans UE1"))
    Ab=float(input("Donner votre moyenne en ... dans UE1"))
    Bb=float(input("Donner votre moyenne en ... dans UE1"))
    Cb=float(input("Donner votre moyenne en ... dans UE1"))
    UE1Note= np.array([Ab,Bb,Cb])
    UE1Coefficient=np.array([Aa,Ba,Ca])
    moyenneUE1=np.average(UE1Note, weights=UE1Coefficient)
    print("Votre moyenne sur l'UE1 est donc de :{:.2f}".format(moyenneUE1))
    Ac=int(input("Donner le coefficient de ... dans UE2"))
    Bc=int(input("Donner le coefficient de ... dans UE2"))
    Cc=int(input("Donner le coefficient de ... dans UE2"))
    Ad=float(input("Donner votre moyenne en ... dans UE2"))
    Bd=float(input("Donner votre moyenne en ... dans UE2"))
    Cd=float(input("Donner votre moyenne en ... dans UE2"))
    UE2Note= np.array([Ad,Bd,Cd])
    UE2Coefficient=np.array([Ac,Bc,Cc])
    moyenneUE2=np.average(UE2Note, weights=UE2Coefficient)
    print("Votre moyenne sur l'UE2 est donc de :{:.2f}".format(moyenneUE2))   
    Ae=int(input("Donner le coefficient de ... dans UE3"))
    Be=int(input("Donner le coefficient de ... dans UE3"))
    Ce=int(input("Donner le coefficient de ... dans UE3"))
    Af=float(input("Donner votre moyenne en ... dans UE3"))
    Bf=float(input("Donner votre moyenne en ... dans UE3"))
    Cf=float(input("Donner votre moyenne en ... dans UE3"))
    UE3Note= np.array([Af, Bf,Cf])
    UE3Coefficient=np.array([Ae,Be,Ce])
    moyenneUE3=np.average(UE3Note, weights=UE3Coefficient)
    print("Votre moyenne sur l'UE3 est donc de :{:.2f}".format(moyenneUE3))
    for i in range (8,10):
        if moyenneUE1>=10 and moyenneUE2>=10 and moyenneUE3>=10:
            print("Vous êtes dans le vert !")
        elif moyenneUE1==i or moyenneUE2==i or moyenneUE3==i:
            print("Vous êtes dans le orange !")
calcul_moyenne()
```

```python
#Version 5

from tkinter import *
from tkinter import ttk
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from PIL import ImageTk
fen=Tk()
fen.geometry("1280x720")
fen.title("Calcul De La Moyenne Des UE - Raphaël Godart | Groupe B")
fram=ttk.Frame(fen)
fram.grid()
matiere=[f"R{i}" for i in range (101, 116)]+[f"SAE{i}" for i in range (11,17)]
coef1=[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2=[4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3=[4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]
saisies=[]
for i, matiere in enumerate(matiere) : 
   mat=ttk.Label(fram, text=matiere)
   mat.grid(column=0 , row=i+1, padx=20, pady=4)
   data=ttk.Entry(fram)
   data.grid(column=1, row=i+1)
   saisies.append(data)
moy={"UE1":0.0, "UE2":0.0, "UE3":0.0}
def recup():
   notes=[float(data.get()) for data in saisies]
   moy["UE1"]=round(np.average(notes, weights=coef1), 2)
   moy["UE2"]=round(np.average(notes, weights=coef2), 2)
   moy["UE3"]=round(np.average(notes, weights=coef3), 2)
   UE1=ttk.Label(fram, text="Moyenne de l'UE1")
   UE1.grid(column=2, row=3, pady=4)
   moy1=ttk.Label(fram, text=moy["UE1"], relief=GROOVE)
   moy1.grid(column =3, row=3, pady=4)
   UE2=ttk.Label(fram, text="Moyenne de l'UE2")
   UE2.grid(column=2, row=5, pady=4)
   moy2=ttk.Label(fram, text=moy["UE2"], relief=GROOVE)
   moy2.grid(column=3, row=5, pady=4)
   UE3=ttk.Label(fram, text="Moyenne de l'UE3")
   UE3.grid(column=2, row=7, pady=4)
   moy3=ttk.Label(fram, text=moy["UE3"], relief=GROOVE)
   moy3.grid(column=3, row=7, pady=4)
def graphique():
    global imagetk
    moyennes=[moy["UE1"], moy["UE2"], moy["UE3"]]
    couleurdugraphique=[]
    for note in moyennes:
        if note>=10:
            couleurdugraphique.append("green")
        elif note>=8:
            couleurdugraphique.append("orange")
        else:
            couleurdugraphique.append("red")
    plt.style.use("dark_background")
    plt.figure("Graphique Des Moyennes")
    plt.ylabel("Moyenne")
    plt.bar(["UE1","UE2","UE3"], moyennes, color=couleurdugraphique)
    plt.ylim(0,20)
    plt.savefig("Graphique_Moyenne_UE.png")
    plt.close()
    images=Image.open("Graphique_Moyenne_UE.png")
    imagetk=ImageTk.PhotoImage(images)
    graph.config(image=imagetk)
btnnotes=Button(fram, text="Valider La Saisie", command=recup)
btnnotes.grid(column=3, row=1, padx=15)
btngraphique=Button(fram, text="Afficher Un Graphique", command=graphique)
btngraphique.grid(column=4, row=1, padx=15)
graph=Label(fram)
graph.grid(column=4, row=3, rowspan=20, padx=15)
fen.mainloop()
```
